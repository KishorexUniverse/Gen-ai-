```python
import os
from getpass import getpass

os.environ["GROQ_API_KEY"] = getpass("Enter your Groq API key: ")
```

```python
%%writefile app.py

import os
import streamlit as st
from groq import Groq

st.set_page_config(
    page_title="Nova AI",
    page_icon="🤖",
    layout="wide"
)

# ---------- CSS ----------
st.markdown("""
<style>

.stApp {
    background:
        radial-gradient(circle at 10% 10%, #312e81 0%, transparent 25%),
        radial-gradient(circle at 90% 20%, #581c87 0%, transparent 25%),
        linear-gradient(135deg, #020617, #111827);
    color: white;
}

.title {
    text-align: center;
    font-size: 48px;
    font-weight: 900;
    background: linear-gradient(90deg, #8b5cf6, #06b6d4, #ec4899);
    -webkit-background-clip: text;
    -webkit-text-fill-color: transparent;
}

.subtitle {
    text-align: center;
    color: #94a3b8;
    margin-bottom: 30px;
}

.user-message {
    background: rgba(139, 92, 246, 0.20);
    border: 1px solid rgba(139, 92, 246, 0.4);
    padding: 15px;
    border-radius: 18px;
    margin: 10px 0;
}

.ai-message {
    background: rgba(6, 182, 212, 0.12);
    border: 1px solid rgba(6, 182, 212, 0.3);
    padding: 15px;
    border-radius: 18px;
    margin: 10px 0;
}

</style>
""", unsafe_allow_html=True)

# ---------- API ----------
api_key = os.getenv("GROQ_API_KEY")

if not api_key:
    st.error("Groq API key not found.")
    st.stop()

client = Groq(api_key=api_key)

# ---------- Sidebar ----------
with st.sidebar:

    st.markdown("## ⚡ Nova AI")

    model = st.selectbox(
        "Choose Model",
        [
            "openai/gpt-oss-20b",
            "openai/gpt-oss-120b"
        ]
    )

    temperature = st.slider(
        "Creativity",
        0.0,
        1.5,
        0.7,
        0.1
    )

    if st.button("🗑️ Clear Chat"):
        st.session_state.messages = []
        st.rerun()

    st.markdown("---")

    st.write("🟢 **Groq Connected**")
    st.write("⚡ Fast inference")
    st.write("💬 Streaming enabled")

# ---------- Chat History ----------
if "messages" not in st.session_state:
    st.session_state.messages = []

# ---------- Header ----------
st.markdown(
    '<div class="title">🤖 Nova AI</div>',
    unsafe_allow_html=True
)

st.markdown(
    '<div class="subtitle">Fast AI Assistant powered by Groq</div>',
    unsafe_allow_html=True
)

# ---------- Display History ----------
for message in st.session_state.messages:

    if message["role"] == "user":

        st.markdown(
            f"""
            <div class="user-message">
            👤 <b>You</b><br><br>
            {message["content"]}
            </div>
            """,
            unsafe_allow_html=True
        )

    else:

        st.markdown(
            f"""
            <div class="ai-message">
            🤖 <b>Nova AI</b><br><br>
            {message["content"]}
            </div>
            """,
            unsafe_allow_html=True
        )

# ---------- Input ----------
prompt = st.chat_input("Ask Nova anything...")

if prompt:

    # Save user message
    st.session_state.messages.append({
        "role": "user",
        "content": prompt
    })

    st.markdown(
        f"""
        <div class="user-message">
        👤 <b>You</b><br><br>
        {prompt}
        </div>
        """,
        unsafe_allow_html=True
    )

    # AI response
    placeholder = st.empty()
    response = ""

    try:

        stream = client.chat.completions.create(
            model=model,
            messages=[
                {
                    "role": "system",
                    "content": """
You are Nova AI, a helpful and intelligent AI assistant.
Give accurate, clear and useful answers.
For programming questions, provide working code.
Explain difficult concepts simply.
"""
                }
            ] + st.session_state.messages,
            temperature=temperature,
            max_tokens=2048,
            stream=True
        )

        for chunk in stream:

            text = chunk.choices[0].delta.content

            if text:
                response += text

                placeholder.markdown(
                    f"""
                    <div class="ai-message">
                    🤖 <b>Nova AI</b><br><br>
                    {response}▌
                    </div>
                    """,
                    unsafe_allow_html=True
                )

        placeholder.markdown(
            f"""
            <div class="ai-message">
            🤖 <b>Nova AI</b><br><br>
            {response}
            </div>
            """,
            unsafe_allow_html=True
        )

        st.session_state.messages.append({
            "role": "assistant",
            "content": response
        })

    except Exception as e:

        st.error(f"Error: {e}")
```
